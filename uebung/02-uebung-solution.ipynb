
## Aufgabe 3 - Phrasenstrukturbaum

### (a) Geben Sie nun für den Satz aus Aufgabe 1 einen vollständigen Phrasenstrukturbaum inklusive Wortart-Label an:

*Fischers Fritz fischt die frischen Fische aus dem Fluss.*



```python
tree = Tree.fromstring("""
(S
    (NP (PROPN Fischers) (PROPN Fritz))
    (VP
        (VP
            (V fischt)
            (NP (DET die) (ADJ frischen) (N Fische))
        )
        (PP 
            (P aus)
            (NP (DET dem) (N Fluss))
        )
    )
)
""")

tree.pretty_print(unicodelines=True)
```

### (b) Geben Sie die entsprechenden Regeln einer kontextfreien Grammatik (CFG) an und überprüfen Sie anschließend, ob der Satz vom CFG-Parser abgeleitet werden kann.


```python
grammar = nltk.CFG.fromstring("""
 # Syntaktische Regeln:
    S -> NP VP
    NP -> PROPN PROPN
    VP -> V NP | VP PP
    NP -> DET ADJ N | DET N
    PP -> P NP

# Lexikalische Regeln:
    PROPN -> "Fischers" | "Fritz"
    V -> "fischt"
    DET -> "die" | "dem"
    N -> "Fische" |"Fluss"
    ADJ -> "frischen"
    P -> "aus"
    """)

sent = "Fischers Fritz fischt die frischen Fische aus dem Fluss"

parser = nltk.ChartParser(grammar)

for tree in parser.parse(sent.split()):
    tree.pretty_print(unicodelines=True)
```

## Aufgabe 4 - Erweiterung Phrasenstrukturgrammatik

#### Gegeben sei folgende einfache kontextfreie Grammatik (CFG):


```python
grammar = nltk.CFG.fromstring("""
    S -> NP VP
    NP -> PROPN
    VP -> V NP
    PROPN -> "Maria" | "Moritz"
    V -> "kennt"
""")
```

### (a) Parsen Sie den unten gegebenen Satz mit Hilfe der Grammatik, indem Sie das folgende Python-Skript ausführen. Erläutern Sie die Funktionsweise des Codes durch Kommentare.

```python
# Der zu parsende Satz:
sent = "Maria kennt Moritz"

# Erzeugen eines Chart-Parsers mit der definierten Grammatik:
parser = nltk.ChartParser(grammar)

# Durchlaufen und Parsen des Satzes:
for tree in parser.parse(sent.split()): #Parser liefert Sequenz zurück, da potentiell mehr als ein Parsebaum möglich (Ambiguität)
    print(tree) # unformatierten Syntaxbaum ausgeben (Klammerausdruck)
    tree.pretty_print(unicodelines=True) # formatierten Syntaxbaum ausgeben
```

### (b) Überlegen Sie zunächst, wie viele Sätze von dieser Grammatik generiert werden können, bevor Sie den folgenden Codeblock ausführen:

```python
for sentence in generate(grammar, depth=5):
    print(' '.join(sentence))
```

### (c) Erweitern Sie die Grammatik um syntaktische sowie lexikalische Regeln für eine adverbiale Ergänzung mit *(sehr) gut*.

#### Testen Sie dazu mit untenstehenden Beispielsätzen, ob Ihre Grammatik den Satz ableitet:

```python
sents = [
    "Maria kennt Moritz gut",
    "Maria kennt Moritz sehr gut",
]
```

```python
grammar = nltk.CFG.fromstring("""
    S -> NP VP
    NP -> PROPN
    VP -> V NP
    PROPN -> "Maria" | "Moritz"
    V -> "kennt"

# HINZUGEFÜGTE REGELN:
    VP -> V NP ADJP
    ADJP -> ADJ | PRT ADJ
    ADJ -> "gut"
    PRT -> "sehr"
""")

parser = nltk.ChartParser(grammar)
for sent in sents:
    trees = list(parser.parse(sent.split()))
    if trees: [tree.pretty_print(unicodelines=True) for tree in trees]
    else: print(f"no parse found for: {sent}")
```

## *Aufgabe 5 - Distributionsanalysen*

### Erläutern Sie die beiden folgenden Programmaufrufe im Hinblick auf das linguistische Prinzip der Distributionsäquivalenz.

### (a)  Mit Hilfe der `similar`-Methode von NLTK können distributionsäquivalente Wörter gesucht werden, also solche, die in gleichen Kontexten auftreten (vgl. https://www.nltk.org/book/ch05.html#using-a-tagger).

```python
from nltk.corpus import brown
text = nltk.Text(word.lower() for word in brown.words())
text.similar('big')
```

```python
# Erläuterung:
# Feststellung von Adjektiven als Klasse distributionsäquivalenter Wörter
# Empirisch begründete Kategorisierung syntaktischer Einheiten (hier Wortart Adjektiv)
```

### (b)  Mit `common_contexts` können wir die exakten syntaktischen Umgebungen ermitteln, die zwei - zuvor mit ihren Wortart-Labels (wort/TAG) kombinierten - Wörter im Korpus miteinander teilen.

```python
tagged_text = nltk.Text(f"{word.lower()}/{tag}" for word, tag in brown.tagged_words(tagset='universal'))
tagged_text.common_contexts(["big/ADJ", "little/ADJ"])
```

```python
# Erläuterung:
# Expliziter Nachweis der geteilten, strukturellen Kontexte zweier spezifischer Wort-Tag-Einheiten
# Ausgabe der exakten syntaktischen Schablonen, in denen die fixierten Wortarten im Korpus austauschbar sind
# Sichtbarmachen typischer grammatikalischer Muster (z. B. das Auftreten von Adjektiven in der Slot-Position zwischen DET und NOUN)
```
