## Aufgabe 4 - Erweiterung Phrasenstrukturgrammatik

#### Gegeben sei folgende einfache kontextfreie Grammatik (CFG):


```python
grammar = nltk.CFG.fromstring("""
    S -> NP VP
    NP -> PROPN
    VP -> V NP
    PROPN -> "Maria" | "Moritz"
    V -> "kennt"
""")
```

### (a) Parsen Sie den unten gegebenen Satz mit Hilfe der Grammatik, indem Sie das folgende Python-Skript ausführen. Erläutern Sie die Funktionsweise des Codes durch Kommentare.

```python
sent = "Maria kennt Moritz"

parser = nltk.ChartParser(grammar)
for tree in parser.parse(sent.split()):
    print(tree)
    tree.pretty_print(unicodelines=True)
```

### (b) Überlegen Sie zunächst, wie viele Sätze von dieser Grammatik generiert werden können, bevor Sie den folgenden Codeblock ausführen:

```python
for sentence in generate(grammar, depth=5):
    print(' '.join(sentence))
```

### (c) Erweitern Sie die Grammatik um syntaktische sowie lexikalische Regeln für eine adverbiale Ergänzung mit *(sehr) gut*.

#### Testen Sie dazu mit untenstehenden Beispielsätzen, ob Ihre Grammatik den Satz ableitet:

```python
sents = [
    "Maria kennt Moritz gut",
    "Maria kennt Moritz sehr gut",
]
```

```python
grammar = nltk.CFG.fromstring("""
    S -> NP VP
    NP -> PROPN
    VP -> V NP
    PROPN -> "Maria" | "Moritz"
    V -> "kennt"

# HINZUGEFÜGTE REGELN:

""")

parser = nltk.ChartParser(grammar)
for sent in sents:
    trees = list(parser.parse(sent.split()))
    if trees: [tree.pretty_print(unicodelines=True) for tree in trees]
    else: print("no parse found for: " + sent) 
```
